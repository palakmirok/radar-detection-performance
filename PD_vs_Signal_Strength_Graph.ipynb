```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

trials = 200000
threshold = 3.0

strengths = np.arange(0, 8.1, 0.5)   # target ki strength
pd_sim = []

for s in strengths:
    x = s + np.random.randn(trials)
    pd_sim.append(np.mean(x > threshold))

pd_theory = norm.sf(threshold - strengths)   # theory wali value

plt.plot(strengths, pd_sim, 'o', label='Simulation')
plt.plot(strengths, pd_theory, '-', label='Theory')
plt.xlabel('Target signal strength')
plt.ylabel('Pd')
plt.title('Pd vs signal strength (threshold = 3)')
plt.grid(True)
plt.legend()
plt.show()
```
