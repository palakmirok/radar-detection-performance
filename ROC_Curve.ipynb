```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

trials = 200000
signal = 2.0                      # target ki strength (ise badal ke bhi dekhna)

noise = np.random.randn(trials)               # sirf noise
target = signal + np.random.randn(trials)     # signal + noise

thresholds = np.linspace(-4, 8, 300)

pfa = [np.mean(noise > t) for t in thresholds]
pd = [np.mean(target > t) for t in thresholds]

plt.plot(pfa, pd, label=f'Signal strength = {signal}')
plt.plot([0, 1], [0, 1], 'k--', label='Random guessing')
plt.xlabel('Pfa')
plt.ylabel('Pd')
plt.title('ROC curve')
plt.grid(True)
plt.legend()
plt.show()
```
